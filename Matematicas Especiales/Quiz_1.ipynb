```python
# Un modelo de lenguaje genera el embedding W = [0.25, -0.3, 0.1, 0.4] para una palabra. Calcula la norma Euclidiana y la norma Manhattan  del vector W.
# To calculate the Euclidean norm and the Manhattan norm of the vector W = [0.25, -0.3, 0.1, 0.4], we can use the following formulas:

w = np.array([0.25, -0.3, 0.1, 0.4])
# Euclidean norm (L2 norm)
euclidean_norm = np.linalg.norm(w)
# Manhattan norm (L1 norm)
manhattan_norm = np.sum(np.abs(w))

print(f"Euclidean norm of W: {euclidean_norm}")
print(f"Manhattan norm of W: {manhattan_norm}")
```

Asuma que un jugador gana $1 con probabilidad 0.4 en cada juego. Por lo tanto, la probabilidad de que el jugador acumule $25, empezando con $10 será



*Datos del problema:*

* Capital inicial i: 10
* Capital meta N: 25
* Probabilidad de ganar (\(p\)): 0.4
* Probabilidad de perder (\(q\)): 0.6 (calculado como \(1 - 0.4\))


```python
# Asuma que un jugador gana $1 con probabilidad 0.4 en cada juego. Por lo tanto, la probabilidad de que el jugador acumule $25, empezando con $10 será

i = 10 # initial amount
n = 25 # target amount
p = 0.4 # probability of winning
q = 1 - p # probability of losing

r = q / p

print(f"Ration: {r}")

probability = (1 - (pow(r, i))) / (1 - (pow(r, n)))

print(f"Probabilidad de que el jugador acumule $25 empezando con $10: {probability}")
```
